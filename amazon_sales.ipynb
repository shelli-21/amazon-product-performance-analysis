# Project: Amazon E-Commerce Product Performance & Pricing Analysis

### 🎯 Objective
An analysis of Amazon product data to understand how pricing strategies, discount depths, and review volumes impact customer satisfaction (ratings) and product discoverability.

### 🚀 Product Insights (TL;DR)
* **The Rating Baseline:** High-volume products maintain a stable rating baseline around 4.0 - 4.5⭐, while low-volume items skew heavily.
* **Discount Elasticity:** High discount percentages (like the 64% and 90% seen below) directly correlate with massive review volumes (24k+ reviews), showing promotions drive acquisition.

```python
# Check average rating and review volume by discount tiers
df['discount_tier'] = pd.cut(df['discount_percentage'], bins=[0, 30, 60, 100], labels=['Low (<30%)', 'Medium (30-60%)', 'High (>60%)'])
df.groupby('discount_tier', observed=False).agg({
    'rating': 'mean',
    'rating_count': 'mean',
    'product_id': 'count'
}).rename(columns={'product_id': 'total_products'})
```

### 📊 Insight: The Discount Paradox
* **Observation:** Products with a **High Discount (>60%)** bring the highest customer traffic, averaging **21,444 reviews** per product.
* **Product Impact:** Deep discounting is highly effective for aggressive user acquisition. However, the average customer rating drops from `4.15` to `4.06`.
* **Recommendation:** While promotions clear inventory and drive traffic, high-discount products need monitoring as they correlate with a slight drop in customer satisfaction.

```python
# Flag popular items with poor customer experience
poor_experience = df[(df['rating'] < 3.8) & (df['rating_count'] > 5000)]
poor_experience[['product_name', 'category', 'rating', 'rating_count', 'discount_percentage']].head()
```

### ⚠️ Product Health Alert: High-Volume Defects
* **Observation:** Multiple heavily discounted products—like the **PTron Tangent Lite Earphones (67% discount)**—maintain massive platform visibility (**83,996 reviews**) despite a low customer satisfaction rating of **3.5⭐**.
* **Product Impact:** Keeping low-quality items pinned to the top via discounts hurts long-term platform trust.
* **Actionable Next Step:** Flag these specific SKUs to the Quality Assurance (QA) and Vendor Management teams for a product health check or supplier review.
